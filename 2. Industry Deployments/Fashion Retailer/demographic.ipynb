```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("results/aldo_demo.csv")
```

```python
filtered_df = df[~df["Status"].isin(["RETURNED", "TIMED-OUT"])]
```

```python
# Count the number of males and females
sex_counts = filtered_df["Sex"].value_counts()

# Display the counts
print("Gender Distribution:")
print(sex_counts)
```

```python
plt.figure(figsize=(8, 8))
plt.pie(sex_counts, labels=sex_counts.index, autopct="%1.1f%%",
        colors=["lightblue", "lightpink"], startangle=90)
plt.title("Marketing Case Study Survey: Gender Distribution of Participants", fontsize=14)
plt.axis("equal")
plt.savefig("results/aldo_gender_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df_clean = filtered_df.copy()
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors="coerce")
average_age = df_clean["Age"].mean()
print(f"Average Age: {average_age:.1f} years")
min_age = df_clean["Age"].min()
max_age = df_clean["Age"].max()
print(f"Age Range: {min_age:.0f} to {max_age:.0f} years")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_clean["Age"], bins=20, edgecolor="black", color="skyblue")
plt.title("Marketing Case Study Survey: Age Distribution of Participants", fontsize=14)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.grid(True, alpha=0.3)
plt.savefig("results/aldo_age_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
country_counts = df_clean["Country of residence"].value_counts()

plt.figure(figsize=(10, 6))
country_counts.plot(kind="bar")
plt.title("Marketing Case Study Survey: Country of Residence Distribution")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.savefig("results/aldo_country_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```
