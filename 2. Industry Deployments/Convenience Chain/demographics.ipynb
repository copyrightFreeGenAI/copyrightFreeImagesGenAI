```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv("results/first.csv")
df2 = pd.read_csv("results/second.csv")
```

```python
combined_df = pd.concat([df1, df2], ignore_index=True)
filtered_df = combined_df[~combined_df["Status"].isin(["RETURNED", "TIMED-OUT"])]
```

```python
len(filtered_df)
```

```python
# Count the number of males and females
sex_counts = filtered_df["Sex"].value_counts()

# Display the counts
print("Gender Distribution:")
print(sex_counts)
```

```python
# Charts below exclude participants with no usable gender value, so they cover
# fewer rows than the status-filtered count printed above.
reported_df = filtered_df[~filtered_df["Sex"].isin(["Prefer not to say", "DATA_EXPIRED"])]
print(f"Status-filtered: {len(filtered_df)} | charted below: {len(reported_df)}")

sex_counts = reported_df["Sex"].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(sex_counts, labels=sex_counts.index, autopct="%1.1f%%",
        colors=["lightblue", "lightpink"], startangle=90)
plt.title("Convenience Retail Survey: Gender Distribution of Participants", fontsize=14)
plt.axis("equal")
plt.savefig("results/circlek_gender_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df_clean = reported_df.copy()
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors="coerce")
average_age = df_clean["Age"].mean()
print(f"Participants charted: {len(df_clean)}")
print(f"Average Age: {average_age:.1f} years")
min_age = df_clean["Age"].min()
max_age = df_clean["Age"].max()
print(f"Age Range: {min_age:.0f} to {max_age:.0f} years")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_clean["Age"], bins=20, edgecolor="black", color="skyblue")
plt.title("Convenience Retail Survey: Age Distribution of Participants", fontsize=14)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.grid(True, alpha=0.3)
plt.savefig("results/circlek_age_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
country_counts = df_clean["Country of residence"].value_counts()

plt.figure(figsize=(10, 6))
country_counts.plot(kind="bar")
plt.title("Convenience Retail Survey: Country of Residence Distribution")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.savefig("results/circlek_country_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```
